Exercise D.3 (Monte Carlo simulations in numpy) Recall Exercise B.5 and implement a version of in_unit_circle that uses numpy arrays, you should take the code from Exercise B.8 as basis and add a new function in_unit_circle_np.

To write this in Python follow these steps:

    With numpy.random.uniform() create a matrix with random numbers.
    Use element wise computation to get . Hint: True is interpreted as 1 and False as 0 when you try to add boolean values.

Once you have the second implementation test it and compare it with the other version.

    Is the accuracy the same? -> Yes
    Have a look at timeit and see which of your two code versions is faster for which N -> Numpy is faster for larger N
    Plot your results, i.e. for different values of the different execution times for the two versions, maybe use a logarithmic axis.
    Are there other plots that might give you an insight for this example.


```python
import numpy as np
import random
import math
import timeit
import matplotlib.pyplot as plt


# --------------------------------------------------
# 1. Original Python implementation
# --------------------------------------------------

def in_unit_circle(N: int) -> int:
    count = 0

    for i in range(N):
        x = random.uniform(0, 1)
        y = random.uniform(0, 1)

        if x**2 + y**2 <= 1:
            count += 1

    return count


# --------------------------------------------------
# 2. NumPy implementation
# --------------------------------------------------

def in_unit_circle_np(N: int) -> int:

    points = np.random.uniform(0, 1, (N, 2))

    inside = points[:, 0]**2 + points[:, 1]**2 <= 1

    return int(np.sum(inside))


# --------------------------------------------------
# 3. Estimate Pi
# --------------------------------------------------

def estimate_pi(N: int, function) -> float:
    return 4 * function(N) / N


# --------------------------------------------------
# 4. Compare accuracy
# --------------------------------------------------

N = 100000

pi_python = estimate_pi(N, in_unit_circle)
pi_numpy = estimate_pi(N, in_unit_circle_np)

error_python = abs(pi_python - math.pi)
error_numpy = abs(pi_numpy - math.pi)

print("MONTE CARLO PI ESTIMATION")
print("------------------------------")
print(f"Real Pi:       {math.pi:.8f}")
print(f"Python Pi:     {pi_python:.8f}")
print(f"NumPy Pi:      {pi_numpy:.8f}")
print()
print(f"Python error:  {error_python:.8f}")
print(f"NumPy error:   {error_numpy:.8f}")


# --------------------------------------------------
# 5. Execution time comparison
# --------------------------------------------------

N_values = [10, 100, 1000, 10000, 100000, 1000000]

times_python = []
times_numpy = []

print("\nEXECUTION TIME COMPARISON")
print("-----------------------------------------------------")
print(f"{'N':>10} {'Python [s]':>15} {'NumPy [s]':>15} {'Speedup':>12}")

for N in N_values:

    t_python = timeit.timeit(
        lambda: in_unit_circle(N),
        number=3
    ) / 3

    t_numpy = timeit.timeit(
        lambda: in_unit_circle_np(N),
        number=3
    ) / 3

    times_python.append(t_python)
    times_numpy.append(t_numpy)

    speedup = t_python / t_numpy

    print(f"{N:>10} {t_python:>15.6f} {t_numpy:>15.6f} {speedup:>11.2f}x")


# --------------------------------------------------
# 6. Plot execution times
# --------------------------------------------------

plt.figure(figsize=(10, 6))

plt.plot(N_values, times_python, "o-", label="Python")
plt.plot(N_values, times_numpy, "s-", label="NumPy")

plt.xscale("log")
plt.yscale("log")

plt.xlabel("Number of points N")
plt.ylabel("Execution time [s]")
plt.title("Monte Carlo Simulation - Execution Time")

plt.legend()
plt.grid(True, which="both", alpha=0.3)
plt.tight_layout()
plt.show()


# --------------------------------------------------
# 7. Plot speedup
# --------------------------------------------------

speedup_values = np.array(times_python) / np.array(times_numpy)

plt.figure(figsize=(10, 6))

plt.plot(N_values, speedup_values, "o-")

plt.xscale("log")

plt.xlabel("Number of points N")
plt.ylabel("Speedup factor")
plt.title("NumPy Speedup Compared to Python")

plt.axhline(y=1, linestyle="--", label="Equal performance")

plt.legend()
plt.grid(True, which="both", alpha=0.3)
plt.tight_layout()
plt.show()


# --------------------------------------------------
# 8. Accuracy comparison for different N
# --------------------------------------------------

N_accuracy = np.unique(np.logspace(2, 6, 20, dtype=int))

errors_python = []
errors_numpy = []

for N in N_accuracy:

    pi_python = estimate_pi(N, in_unit_circle)
    pi_numpy = estimate_pi(N, in_unit_circle_np)

    errors_python.append(abs(pi_python - math.pi))
    errors_numpy.append(abs(pi_numpy - math.pi))


# --------------------------------------------------
# 9. Plot accuracy
# --------------------------------------------------

plt.figure(figsize=(10, 6))

plt.plot(N_accuracy, errors_python, "o-", label="Python")
plt.plot(N_accuracy, errors_numpy, "s-", label="NumPy")

plt.xscale("log")
plt.yscale("log")

plt.xlabel("Number of points N")
plt.ylabel("Absolute error")
plt.title("Monte Carlo Simulation - Accuracy")

plt.legend()
plt.grid(True, which="both", alpha=0.3)
plt.tight_layout()
plt.show()


# --------------------------------------------------
# 10. Final conclusions
# --------------------------------------------------

print("\nCONCLUSIONS")
print("-----------------------------------------------------")
print("1. Both implementations have similar statistical accuracy.")
print("2. NumPy is generally faster for large N.")
print("3. Both implementations have approximately O(N) time complexity.")
print("4. The typical Monte Carlo error decreases proportionally to 1/sqrt(N).")
print("5. Vectorization makes NumPy more efficient than Python loops.")
```

Exercise D.4 (Predator Prey simulation) The Lotka-Volterra Equations 

```python
import numpy as np
import matplotlib.pyplot as plt


# ============================================================
# EXERCISE D.4 - PREDATOR PREY SIMULATION
# Lotka-Volterra Equations
# ============================================================


# ------------------------------------------------------------
# 1. PREY CLASS
# ------------------------------------------------------------

class Prey:

    def __init__(self, alpha, beta, population):

        # Public parameter
        self.alpha = alpha

        # Private parameter
        self.__beta = None
        self.set_beta(beta)

        # Current population
        self.population = population

    # Getter
    def get_beta(self):
        return self.__beta

    # Setter with validation
    def set_beta(self, beta):

        if isinstance(beta, bool) or not isinstance(
            beta, (int, float, np.number)
        ):
            raise TypeError("Beta must be a number")

        if not np.isfinite(beta) or beta <= 0:
            raise ValueError("Beta must be positive and finite")

        self.__beta = beta


# ------------------------------------------------------------
# 2. PREDATOR CLASS
# ------------------------------------------------------------

class Predator:

    def __init__(self, delta, gamma, population):

        # Public parameter
        self.delta = delta

        # Private parameter
        self.__gamma = None
        self.set_gamma(gamma)

        # Current population
        self.population = population

    # Getter
    def get_gamma(self):
        return self.__gamma

    # Setter with validation
    def set_gamma(self, gamma):

        if isinstance(gamma, bool) or not isinstance(
            gamma, (int, float, np.number)
        ):
            raise TypeError("Gamma must be a number")

        if not np.isfinite(gamma) or gamma <= 0:
            raise ValueError("Gamma must be positive and finite")

        self.__gamma = gamma


# ------------------------------------------------------------
# 3. SIMULATION CLASS
# ------------------------------------------------------------

class Simulation:

    def __init__(self, prey, predator):

        self.prey = prey
        self.predator = predator

    # Calculate derivatives
    def __derivatives(self):

        x = self.prey.population
        y = self.predator.population

        alpha = self.prey.alpha
        beta = self.prey.get_beta()

        delta = self.predator.delta
        gamma = self.predator.get_gamma()

        dx = alpha * x - beta * x * y
        dy = delta * x * y - gamma * y

        return dx, dy

    # Forward Euler integration
    def __step(self, dt):

        dx, dy = self.__derivatives()

        # Update both populations simultaneously
        x_new = self.prey.population + dt * dx
        y_new = self.predator.population + dt * dy

        self.prey.population = x_new
        self.predator.population = y_new

    # Public simulation method
    def simulate(self, t1, dt):

        if t1 <= 0 or dt <= 0:
            raise ValueError("t1 and dt must be positive")

        # Number of steps
        steps = int(np.ceil(t1 / dt))

        # Time array
        time = np.linspace(0, t1, steps + 1)

        # Arrays to store populations
        x_values = np.zeros(steps + 1)
        y_values = np.zeros(steps + 1)

        # Initial populations
        x_values[0] = self.prey.population
        y_values[0] = self.predator.population

        # Forward Euler integration
        for i in range(steps):

            # Actual time step
            current_dt = time[i + 1] - time[i]

            self.__step(current_dt)

            # Store results
            x_values[i + 1] = self.prey.population
            y_values[i + 1] = self.predator.population

        return time, x_values, y_values


# ============================================================
# 4. INITIAL PARAMETERS
# ============================================================

alpha = 1.0
beta = 0.1

gamma = 0.5
delta = 0.02

x0 = 100
y0 = 20

# Simulation settings
t1 = 50
dt = 0.01


# ============================================================
# 5. CREATE OBJECTS
# ============================================================

prey = Prey(
    alpha=alpha,
    beta=beta,
    population=x0
)

predator = Predator(
    delta=delta,
    gamma=gamma,
    population=y0
)

simulation = Simulation(prey, predator)


# ============================================================
# 6. RUN SIMULATION
# ============================================================

time, x, y = simulation.simulate(t1, dt)


# ============================================================
# 7. PRINT RESULTS
# ============================================================

print("LOTKA-VOLTERRA SIMULATION")
print("------------------------------------")

print(f"Alpha: {prey.alpha}")
print(f"Beta:  {prey.get_beta()}")

print(f"Gamma: {predator.get_gamma()}")
print(f"Delta: {predator.delta}")

print()
print(f"Initial prey population:     {x0}")
print(f"Initial predator population: {y0}")

print()
print(f"Simulation time: {t1}")
print(f"Time step:       {dt}")

print()
print(f"Final prey population:     {x[-1]:.2f}")
print(f"Final predator population: {y[-1]:.2f}")


# ============================================================
# 8. PLOT POPULATION DEVELOPMENT
# ============================================================

plt.figure(figsize=(12, 6))

plt.plot(
    time,
    x,
    label="Prey",
    linewidth=2
)

plt.plot(
    time,
    y,
    label="Predator",
    linewidth=2
)

plt.xlabel("Time")
plt.ylabel("Population")

plt.title("Lotka-Volterra - Predator Prey Simulation")

plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()


# ============================================================
# 9. PHASE SPACE PLOT
# ============================================================

plt.figure(figsize=(8, 7))

plt.plot(
    x,
    y,
    linewidth=2
)

plt.scatter(
    x[0],
    y[0],
    s=100,
    label="Start"
)

plt.scatter(
    x[-1],
    y[-1],
    s=100,
    label="End"
)

plt.xlabel("Prey Population")
plt.ylabel("Predator Population")

plt.title("Lotka-Volterra - Phase Space")

plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()


# ============================================================
# 10. TEST GETTER AND SETTER
# ============================================================

print("\nGETTER / SETTER TEST")
print("------------------------------------")

print("Original beta:", prey.get_beta())

prey.set_beta(0.15)

print("Updated beta: ", prey.get_beta())

print()
print("Original gamma:", predator.get_gamma())

predator.set_gamma(0.6)

print("Updated gamma: ", predator.get_gamma())


# ============================================================
# 11. TEST INVALID PARAMETERS
# ============================================================

print("\nVALIDATION TEST")
print("------------------------------------")

try:
    prey.set_beta(-0.1)

except ValueError as error:
    print("Validation successful:", error)

try:
    predator.set_gamma("invalid")

except TypeError as error:
    print("Validation successful:", error)
```

Exercise D.6 (Advanced pandas)  

```python
# ============================================================
# EXERCISE D.6 – ADVANCED PANDAS
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


# ============================================================
# PART 1 – APPLY FUNCTION
# ============================================================

print("PART 1 – AGE CATEGORIES")
print("----------------------------------------")

# Create DataFrame
data = pd.DataFrame({
    "Age": [11, 16, 30, 51]
})

# Function to classify age
def age_category(age):

    if age < 13:
        return "Child"

    elif age < 18:
        return "Teenager"

    else:
        return "Adult"


# Apply function
data["Category"] = data["Age"].apply(age_category)

print(data)


# ============================================================
# PART 2 – DEATH VALLEY DATASET
# ============================================================


# ------------------------------------------------------------
# 1. LOAD DATASET
# ------------------------------------------------------------

print("\nLOAD DATASET")
print("----------------------------------------")

# GitHub raw CSV
url = (
    "https://raw.githubusercontent.com/"
    "Malekai/Downloading-Data/master/"
    "death_valley_2014.csv"
)

# Load data
df = pd.read_csv(url)

# Remove additional spaces from column names
df.columns = df.columns.str.strip()

print(df.head())

print("\nColumns:")
print(df.columns.tolist())


# ------------------------------------------------------------
# 2. CONVERT PST TO DATETIME
# ------------------------------------------------------------

df["PST"] = pd.to_datetime(df["PST"])

print("\nDATETIME CONVERSION")
print("----------------------------------------")

print(df[["PST"]].head())


# ------------------------------------------------------------
# 3. CONVERT TEMPERATURES FROM F TO C
# ------------------------------------------------------------

# Select all temperature columns
temp_columns = [
    "Max TemperatureF",
    "Mean TemperatureF",
    "Min TemperatureF"
]

# Create new DataFrame
temperature_df = df[temp_columns].copy()

# Convert Fahrenheit to Celsius
temperature_df = (temperature_df - 32) * 5 / 9

# Rename columns
temperature_df.columns = [
    "Max TemperatureC",
    "Mean TemperatureC",
    "Min TemperatureC"
]

print("\nTEMPERATURES IN CELSIUS")
print("----------------------------------------")

print(temperature_df.head())


# ------------------------------------------------------------
# 4. MERGE DATAFRAMES
# ------------------------------------------------------------

# Merge temperature data with original data
df = pd.concat(
    [df, temperature_df],
    axis=1
)

print("\nMERGED DATAFRAME")
print("----------------------------------------")

print(df.head())


# ------------------------------------------------------------
# 5. REPEAT CONVERSION USING APPLY
# ------------------------------------------------------------

# Fahrenheit to Celsius function
def fahrenheit_to_celsius(value):

    return (value - 32) * 5 / 9


# Apply function to temperature columns
temperature_apply = (
    df[temp_columns]
    .apply(fahrenheit_to_celsius)
)

temperature_apply.columns = [
    "Max TemperatureC_apply",
    "Mean TemperatureC_apply",
    "Min TemperatureC_apply"
]

print("\nTEMPERATURE CONVERSION USING APPLY")
print("----------------------------------------")

print(temperature_apply.head())


# ------------------------------------------------------------
# 6. CORRECT JANUARY TEMPERATURE ERROR
# ------------------------------------------------------------

# Assume January measurements are 1% too high

january = df["PST"].dt.month == 1

celsius_columns = [
    "Max TemperatureC",
    "Mean TemperatureC",
    "Min TemperatureC"
]

# Correct January values by dividing by 1.01
df.loc[january, celsius_columns] = (
    df.loc[january, celsius_columns] / 1.01
)

print("\nJANUARY TEMPERATURE CORRECTION")
print("----------------------------------------")

print(
    df.loc[january, ["PST"] + celsius_columns].head()
)


# ============================================================
# 7. PIPE FUNCTIONS
# ============================================================


# ------------------------------------------------------------
# FUNCTION 1 – MONTHLY HUMIDITY NORMALIZATION
# ------------------------------------------------------------

def normalize_humidity(data):

    data = data.copy()

    # Monthly mean humidity
    monthly_mean = (
        data.groupby(data["PST"].dt.month)
        ["Mean Humidity"]
        .transform("mean")
    )

    # Normalize humidity using monthly mean
    data["Humidity_Normalized"] = (
        data["Mean Humidity"] / monthly_mean
    )

    return data


# ------------------------------------------------------------
# FUNCTION 2 – WIND DIRECTION CONVERSION
# ------------------------------------------------------------

def convert_wind_direction(data):

    data = data.copy()

    # Convert 0...360 to -180...180
    data["WindDirDegrees"] = (
        (data["WindDirDegrees"] + 180) % 360
    ) - 180

    return data


# ------------------------------------------------------------
# FUNCTION 3 – DATE TO MONTH AND DAY
# ------------------------------------------------------------

def convert_date(data):

    data = data.copy()

    # Keep only month and day
    data["PST"] = data["PST"].dt.strftime("%m-%d")

    return data


# ------------------------------------------------------------
# 8. APPLY PIPELINE
# ------------------------------------------------------------

df_processed = (
    df
    .pipe(normalize_humidity)
    .pipe(convert_wind_direction)
    .pipe(convert_date)
)

print("\nPROCESSED DATAFRAME")
print("----------------------------------------")

print(
    df_processed[
        [
            "PST",
            "Mean Humidity",
            "Humidity_Normalized",
            "WindDirDegrees"
        ]
    ].head()
)


# ============================================================
# 9. PLOTS
# ============================================================


# ------------------------------------------------------------
# PLOT 1 – TEMPERATURE DEVELOPMENT
# ------------------------------------------------------------

plt.figure(figsize=(12, 6))

plt.plot(
    df["PST"],
    df["Max TemperatureC"],
    label="Max Temperature"
)

plt.plot(
    df["PST"],
    df["Mean TemperatureC"],
    label="Mean Temperature"
)

plt.plot(
    df["PST"],
    df["Min TemperatureC"],
    label="Min Temperature"
)

plt.xlabel("Date")
plt.ylabel("Temperature [°C]")

plt.title("Death Valley 2014 – Temperature")

plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# PLOT 2 – HUMIDITY NORMALIZATION
# ------------------------------------------------------------

plt.figure(figsize=(12, 6))

plt.plot(
    df["PST"],
    df_processed["Humidity_Normalized"],
    label="Normalized Humidity"
)

plt.axhline(
    y=1,
    linestyle="--",
    label="Monthly Mean"
)

plt.xlabel("Date")
plt.ylabel("Normalized Humidity")

plt.title("Death Valley 2014 – Normalized Humidity")

plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# PLOT 3 – WIND DIRECTION
# ------------------------------------------------------------

plt.figure(figsize=(12, 6))

plt.scatter(
    df["PST"],
    df_processed["WindDirDegrees"],
    s=12
)

plt.xlabel("Date")
plt.ylabel("Wind Direction [°]")

plt.title("Death Valley 2014 – Wind Direction")

plt.ylim(-180, 180)

plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()


# ============================================================
# 10. FINAL OUTPUT
# ============================================================

print("\nFINAL DATAFRAME")
print("----------------------------------------")

print(
    df_processed[
        [
            "PST",
            "Mean TemperatureC",
            "Humidity_Normalized",
            "WindDirDegrees"
        ]
    ].head(10)
)

print("\nEXERCISE D.6 COMPLETED")
```
